```python
# Path to .env file in Google Drive
env_path = "/content/drive/My Drive/DL/project/accesstoken.env"

# Load .env file
load_dotenv(env_path)

# Access variables
ACCESS_TOKEN = os.getenv("ACCESS_TOKEN")
IG_BUSINESS_ID = os.getenv("INSTAGRAM_BUSINESS_ID")

print("Instagram Business ID:", IG_BUSINESS_ID)
print("access token",ACCESS_TOKEN)
```

```python
BASE_URL = "https://graph.facebook.com/v18.0"
```

```python
import requests
def get_instagram_media():
    url = f"{BASE_URL}/{IG_BUSINESS_ID}/media"
    params = {
        "fields": "id,caption,media_type,media_url,permalink,timestamp",
        "access_token": ACCESS_TOKEN
    }
    response = requests.get(url, params=params)
    return response.json()

# Fetch and print Instagram posts
media_data = get_instagram_media()
print(media_data)
```
